```python
from langgraph.graph import StateGraph, START, END
from typing import TypedDict
from dotenv import load_dotenv
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
llm = HuggingFaceEndpoint(
    repo_id="deepseek-ai/DeepSeek-V3-0324",
    task="text-generation",
)

model = ChatHuggingFace(llm=llm)
```

```python
# create a state
class LLMState(TypedDict):

    question: str
    answer: str
```

```python
def llm_qa(state: LLMState) -> LLMState:
    question = state["question"]

    # form a prompt for the LLM
    prompt = f"Answer the following question: {question}"

    # get the answer from the LLM
    answer = model.invoke(prompt).content

    # return the new state with the answer
    state["answer"] = answer
    return state
```

```python
# create graph
graph = StateGraph(LLMState)

#add nodes to the graph
graph.add_node('llm_qa', llm_qa)

#add edges to the graph
graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa', END)

#compile the graph
workflow = graph.compile()
```

```python
#execute the workflow with initial state
initial_state = {'question': 'How far is the moon from the earth?'}

final_state = workflow.invoke(initial_state)

print(final_state['answer'])  # print the final state with the answer
```
